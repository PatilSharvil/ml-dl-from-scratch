```python
import pandas as pd

# Series -> pandas series is a one dimentional array like object that can hold any data type

data = [1,2,3,4,5]
series = pd.Series(data)
print(series)
print(type(series)) 
```

```python
# create a series from dictionary 
data = {'a':1, 'b':2, 'c':3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10,20,30]
index = ['a','b','c']
pd.Series(data, index=index)
```

```python
# Dataframe
# Creating a dataframe from a dictionary of list

data = {
    'Name':['Rohan', 'Raghav', 'Raj'],
    'Age':[25,30,45],
    'City':['Mumbai', 'Pune', 'Banglore']
}

df = pd.DataFrame(data)
```
