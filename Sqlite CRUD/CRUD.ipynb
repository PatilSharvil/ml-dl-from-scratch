SQL is a standard language for managing and manipulating relational databases. SQlite is a self contained, serverless and zero config database engine that is widely used for embedded database systems. below are the basics of SQL and SQLite including creating databases, tables and performing various SQL operations 

```python
import sqlite3 
```

```python
# Connect to an SQLite databse (or create it if it doesn't exist)
connection = sqlite3.connect('example.db')
```

```python
cursor = connection.cursor()
```

```python
cursor.execute ( '''
    create table if not exists employees (
        id Integer primary key,
        name text not null,
        age Integer,
        dept text
    )
    '''
)

# Commit the changes 
connection.commit()
```

```python
cursor.execute ( '''
    select * from employees 
    '''
)
```

```python
# insert the data into sqlite table : 

cursor.execute(
    '''
    insert into employees(name, age, dept)
    values('Sharvil', 21, 'Computer Engineer')
    '''
)

cursor.execute(
    '''
    insert into employees(name, age, dept)
    values('Rohit', 25, 'AI/ML Engineer')
    '''
)

connection.commit()
```

```python
# Query the data from the sqlite data or from the table itself : 

cursor.execute(
    '''
    select * from employees
    '''
)
rows = cursor.fetchall()

# printing the data : 
for row in rows : 
    print(row)
```

```python
# Update the data in the table : 
cursor.execute(
    '''
    update employees
    set age = 24 where name = 'Rohit'
    '''
)

connection.commit()
```

```python
cursor.execute(
    '''
    select * from employees
    '''
)
rows = cursor.fetchall()

# printing the data : 
for row in rows : 
    print(row)
```

```python
# Delete the data from the table : 
cursor.execute(
    '''
    delete from employees where name = 'Rohit'
    '''
)

connection.commit()
```

```python
cursor.execute(
    '''
    select * from employees
    '''
)
rows = cursor.fetchall()

# printing the data : 
for row in rows : 
    print(row)
```

```python
# many execution : 
emp_data = [
    ('John', 28, 'HR'),
    ('Alice', 30, 'Finance'),
    ('Bob', 26, 'IT')
]

cursor.executemany(
    '''
    insert into employees(name, age, dept)
    values(?,?,?)
    ''', emp_data
)
```

```python
cursor.execute(
    '''
    select * from employees
    '''
)

rows = cursor.fetchall()

rows
```

```python
# close the connection : 
connection.close()
```
