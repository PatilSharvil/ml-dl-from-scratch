```python
nested_tuple = ((1,2,3), (4,5,6), ("Hello", 'C', 3.14), (True,True))
```

```python
print(nested_tuple)
```

```python
for sub_tpl in nested_tuple : 
    print(sub_tpl)
```

```python
for sub_tpl in nested_tuple : 
    for tpl in sub_tpl : 
        print(tpl)
```
