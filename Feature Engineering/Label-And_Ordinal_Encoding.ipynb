```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.DataFrame({
    'color':['red', 'blue', 'green', 'green', 'red', 'blue']
})
```

```python
df.head()
```

```python
lbl_encoder = LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['color']])
```

```python
lbl_encoder.transform([['red']])
```

```python
lbl_encoder.transform([['green']])
```

```python
lbl_encoder.transform([['blue']])
```

```python
# Ordinal Encoding -> when the work is nominal then we can use the lable encoding but when we have to give some ranks in that case we use the ordinal encoding 
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.DataFrame(
    {
        'size' : ['small', 'medium', 'large', 'medium', 'small', 'large']
    }
)
```

```python
df
```

```python
# using the instance of ordinal encoder that is encoder in our case using that we will apply fit_transform to it 
encoder = OrdinalEncoder(categories=[['small', 'medium', 'large']]) # the categories decide what to give less value and what to give large value and that is based on the order of them we write 
# small will be having less value, medium will be having less than large, and large will be having the largest value or rank here in our case
```

```python
encoder.fit_transform(df[['size']])
```
