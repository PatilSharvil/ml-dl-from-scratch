```python
# Mathematical operations : 
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}
```

```python
# Union : 
set1.union(set2)
```

```python
union_set = set1.union(set2)
```

```python
# Intersection : 
intersection_set = set1.intersection(set2)
```
