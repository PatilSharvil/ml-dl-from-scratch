```python
# lambda functions are small anonymous functions defined using the lambda keyword. they can have any number of arguments but only one expression. They are commonly used for short operations or as a argument to higher order function
```

syntax : 

lambda args : expression

```python
addition = lambda a,b : a + b
```

```python
addition(5,4)
```

```python
addition(3,7)
```

```python
is_even = lambda num : num % 2 == 0
```

```python
is_even(12)
```

```python
is_even(13)
```

```python
numbers = [1,2,3,4,5,6]
```

```python
list(map(lambda x:x**2, numbers))
```
