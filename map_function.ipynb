```python
def sqr(num) : 
    return num**2

nums = [1,2,3,4]
list(map(sqr, nums))
```

```python
# map multiple iterables : 
nums1 = [1,2,3]
nums2 = [4,5,6]

added_list = list(map(lambda x,y:x + y, nums1, nums2))
added_list
```

```python
# convert a list of strings to interger 
lst = ['1','2','3']
ints = list(map(int, lst))
```
