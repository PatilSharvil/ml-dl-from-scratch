```python
with open('example.txt', 'r') as file : 
    content = file.read()
    print(content)
```

```python
# printing line by line
with open('example.txt', 'r', newline='') as file : 
    for line in file : 
        print(line.strip())
```

```python
# writing a file
with open('example.txt', 'w') as file : 
    file.write('This is a new line written into the file')
    file.write('hope this doesnt creates any mess around')
```

```python
with open('example.txt', 'a') as file: 
    file.write('\nappending data\n')
```

```python
with open('example.txt', 'r') as file: 
    content = file.read()
    print(content.strip())
```

```python
# wrting a list of lines to a file
lines = ['hello world\n', 'second line\n', 'third line\n']
with open('example.txt', 'w') as file: 
    file.writelines(lines)
with open('example.txt', 'r') as file: 
    data = file.read()
    print(data.strip())
```
