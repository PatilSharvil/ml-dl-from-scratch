```python
lst = [1, 7, 3, 10, 5]
max_diff = 0
for i in range(1, len(lst)) : 
    if abs(lst[i - 1] - lst[i]) > max_diff : 
        max_diff = abs(lst[i - 1] - lst[i])
print(f"Max : {max_diff}")
```
