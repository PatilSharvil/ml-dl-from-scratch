```python

class Car: 
    def __init__(self, windows, doors):
        self.windows = windows
        self.doors = doors
    
    def get_doors(self):
        return self.windows

car = Car(4,5)
print(car.get_doors())
```

```python
class Bank: 
    def __init__(self, owner, balance):
        self.owner=owner
        self.balance=balance
    
    def deposit(self, amount): 
        self.balance = self.balance + amount
        print(f"{amount} deposited\nNew balance: {self.balance}")
    
    def withdraw(self, amount): 
        if amount > self.balance : 
            print(f"not sufficient balance")
        else : 
            self.balance -= amount
    
    def get_balance(self): 
        print(f"Balance : {self.balance}")

user = Bank('Sharvil', 0)
user.withdraw(20)
user.get_balance()
user.deposit(400)
user.get_balance()
user.withdraw(100)
user.get_balance()
  
```
