```python
import random

class School: 
    def __init__(self, name, roll):
        self.name = name # public variable
        self.roll = roll # public variable
        self.__secret = random.randint(1,10) # private variable
    
    def get_student_secret(self): 
        return self.__secret
    
std = School('sharvil', 3003)
print(std.get_student_secret())
print(std.name)
print(std.roll)
```

```python
class Car: 
    def __init__(self, color, windows, doors):
        self._color = color # protected variable
        self.windows = windows # public variable
        self._doors = doors # protected variable
    
    def specs(self): 
        print(f'color: {self._color}\nwindows: {self.windows}\ndoors: {self._doors}')
    
class Tesla(Car): 
    def __init__(self, color, windows, doors, is_selfdriving):
        super().__init__(color, windows, doors)
        self.is_selfdriving = is_selfdriving
        
    def change_color(self, color): 
        super()._color = color
        print(f'changed color: {super()._color}')   
```

```python
car = Tesla('red', 4,4,True)
```
