```python
class Car: 
    def __init__(self, windows, doors, enginetype): 
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype
    
    def specifications(self): 
        print(f"windows : {self.windows}")
        print(f"doors : {self.doors}")
        print(f"engine type : {self.enginetype}")

class Tesla(Car): 
    def __init__(self, windows, doors, enginetype, is_selfdriving):
        super().__init__(windows, doors, enginetype)
        self.self_driving = is_selfdriving
    
car = Tesla(4,4,'electrical', True)
car.specifications()
car.self_driving
```

```python
## Multiple Inheritance
class Hardware: 
    def __init__(self, micro_controller):
        self.mc = micro_controller
    
class Software: 
    def __init__(self, os):
        self.Os = os

class Pc(Hardware, Software): 
    def __init__(self, micro_controller, os, company):
        Hardware.__init__(self, micro_controller)
        Software.__init__(self, os)
        self.company = company
        
pc = Pc('gigabyte', 'windows', 'acer')
```
