```python
class Animal: 
    def speak(self): 
        print(f'animal speaks')

class Dog(Animal): 
    def speak(self):
        print(f'woof')

class Cat(Dog): 
    def speak(self): 
        print(f'meow meow')
        
cat = Cat()
```

```python
cat.speak()
```

```python
dog = Dog()
```

```python
dog.speak()
```
