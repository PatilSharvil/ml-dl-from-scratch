```python
def iseven(num):
    return num % 2 == 0

print(f"number is even ? : {iseven(4)}")
```

```python
# function with multiple parameters 
def full_name(first_name, middle_name, last_name) : 
    return first_name + middle_name + last_name

print(f"the full name of this person is : {full_name("sharvil ", "sunil ", "patil")}")
```

```python
# default parameter
def greet(name= "guest") : 
    print(f"hello {name}, welcome to paradise")

greet("sharvil")
greet()
```

```python
# variable length arguments : 
# positional arguments

def print_numbers(*args) : 
    for num in args : 
        print(num, end=" ")

print_numbers(1,2,3,4,5,6,"Sharvil", 3.14)
```

```python
# keyword arguments : all the paramenters will be in key value pairs only
def print_details(**kwargs) : 
    for key, value in kwargs.items() : 
        print(f"{key} : {value}")
        
print_details(name="sharvil", age= 20, city= "Sangli")
```
