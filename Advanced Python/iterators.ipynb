# Iterators in Python

An iterator is an object that contains a countable number of values. An iterator is an object that can be iterated upon, meaning that you can traverse through all the values. Technically, in Python, an iterator is an object which implements the iterator protocol, which consist of the methods __iter__() and __next__().

## 1. Basic Concept of Iterators

```python
# Example of using iter() and next()
my_list = [1, 2, 3, 4, 5]

# Create an iterator object
my_iter = iter(my_list)

print("Using next() function:")
print(next(my_iter))  # Output: 1
print(next(my_iter))  # Output: 2
print(next(my_iter))  # Output: 3
print(next(my_iter))  # Output: 4
print(next(my_iter))  # Output: 5
# print(next(my_iter))  # This would raise StopIteration exception
```

## 2. Using for loop with iterators

```python
# A for loop actually creates an iterator object and executes the next() method for each loop
my_list = [6, 7, 8, 9, 10]

for item in my_list:
    print(item)
```

## 3. Creating a custom iterator class

```python
# Custom iterator class that counts from 1 to a given number
class Count:
    def __init__(self, max):
        self.max = max
        self.num = 0
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.num >= self.max:
            raise StopIteration
        else:
            self.num += 1
            return self.num

# Using the custom iterator
counter = Count(5)

for num in counter:
    print(num)
```

## 4. Another custom iterator example - Fibonacci sequence

```python
# Iterator that generates Fibonacci numbers up to a limit
class Fibonacci:
    def __init__(self, max_value):
        self.max_value = max_value
        self.a, self.b = 0, 1
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.a > self.max_value:
            raise StopIteration
        else:
            fib = self.a
            self.a, self.b = self.b, self.a + self.b
            return fib

# Using the Fibonacci iterator
fibonacci = Fibonacci(50)

print("Fibonacci numbers up to 50:")
for num in fibonacci:
    print(num, end=" ")
print()
```

## 5. Infinite iterator example

```python
# An infinite iterator that keeps counting
class InfiniteCounter:
    def __init__(self, start=0, step=1):
        self.num = start
        self.step = step
    
    def __iter__(self):
        return self
    
    def __next__(self):
        current = self.num
        self.num += self.step
        return current

# Using infinite iterator (be careful in practice!)
inf_counter = InfiniteCounter(start=10, step=5)

print("First 5 numbers from infinite counter:")
for i, num in enumerate(inf_counter):
    if i >= 5:  # Stop after 5 iterations to prevent infinite loop
        break
    print(num)
```

## 6. Using iter() with a callable and sentinel

```python
# Using iter() with a callable and sentinel
import random

# Generate random numbers until we get 5
def random_number():
    return random.randint(1, 10)

# Creates an iterator that calls random_number() until it returns 5
random_iter = iter(random_number, 5)

print("Random numbers until we hit 5:")
for num in random_iter:
    print(num)
```

## 7. Iterators vs iterables

```python
# Demonstrating the difference between iterables and iterators
my_list = [1, 2, 3, 4]

# my_list is iterable but not an iterator
print(f"my_list is iterable: {hasattr(my_list, '__iter__')}" )
print(f"my_list is iterator: {hasattr(my_list, '__next__')}")

# Convert it to an iterator
my_iterator = iter(my_list)
print(f"my_iterator is iterable: {hasattr(my_iterator, '__iter__')}")
print(f"my_iterator is iterator: {hasattr(my_iterator, '__next__')}")

# An iterator is also iterable
print(f"my_iterator == iter(my_iterator): {my_iterator is iter(my_iterator)}")  # True - iterators are their own iterators
```

## 8. Practical example: File reading with iterators

```python
# Demonstrating file reading using iterators

# Create a sample text file first
with open('sample.txt', 'w') as f:
    f.write('Line 1\nLine 2\nLine 3\nLine 4\n')

# Reading file line by line using iterator
with open('sample.txt', 'r') as f:
    for line_num, line in enumerate(f):  # file objects are iterators!
        print(f"Line {line_num + 1}: {line.strip()}")
```

## 9. Generator expressions vs iterators

```python
# Iterator from range (range is actually an iterable that returns an iterator)
numbers_iter = iter(range(5))
print("Numbers from range iterator:", list(numbers_iter))

# Generator expression (creates an iterator)
squares_gen = (x**2 for x in range(5))
print("Squares from generator expression:", list(squares_gen))
```

## 10. Exception handling with iterators

```python
# Properly handling StopIteration exception
my_list = [1, 2, 3]
my_iter = iter(my_list)

while True:
    try:
        item = next(my_iter)
        print(item)
    except StopIteration:
        print("Iterator exhausted")
        break
```

## Key Points to Remember:

1. **Iterable**: An object that can return an iterator (has __iter__ method)
2. **Iterator**: An object that produces values (has both __iter__ and __next__ methods)
3. **For loops** automatically handle the StopIteration exception
4. **Iterators are consumed**, once exhausted they need to be recreated
5. **Memory efficient**, as they produce values one at a time rather than storing all values in memory
